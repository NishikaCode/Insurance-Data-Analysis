```python
#Dataset load karo

import pandas as pd
import sqlite3

file_path = r"C:\Users\nishika tendane\Downloads\archive (11)\insurance_data.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python
#Dataset ko inspect karo

# Step 3: Explore the Dataset

print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
#Check and clean column names

# Step 4: Clean Column Names

df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace(r"[^a-z0-9_]", "", regex=True)
)

print("Cleaned Column Names:")
print(df.columns.tolist())
```

```python
#Check missing values

# Step 5: Analyze Missing Values

missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values[missing_values > 0])

print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
#Remove duplicate records

# Step 6: Remove Duplicates

duplicates_before = df.duplicated().sum()

df = df.drop_duplicates().copy()

print("Duplicates Found:", duplicates_before)
print("Dataset Shape After Removing Duplicates:", df.shape)
```

```python
#Handle missing values

# Step 7: Handle Missing Values

for col in df.columns:
    if df[col].isnull().any():
        if pd.api.types.is_numeric_dtype(df[col]):
            df[col] = df[col].fillna(df[col].median())
        else:
            mode_values = df[col].mode()
            if not mode_values.empty:
                df[col] = df[col].fillna(mode_values.iloc[0])
            else:
                df[col] = df[col].fillna("Unknown")

print("Missing Values After Cleaning:")
print(df.isnull().sum().sum())
```

```python
#Basic statistical analysis

# Step 8: Statistical Summary

print("Numerical Column Summary:")
display(df.describe())

print("\nCategorical Column Summary:")
display(df.describe(include="object"))
```

```python
#Create a SQL database

# Step 9: Load Data into SQLite

conn = sqlite3.connect("insurance_analysis.db")

df.to_sql(
    "insurance",
    conn,
    if_exists="replace",
    index=False
)

print("Dataset loaded into SQL database successfully!")

# Step 9: Load Data into SQLite

conn = sqlite3.connect("insurance_analysis.db")

df.to_sql(
    "insurance",
    conn,
    if_exists="replace",
    index=False
)

print("Dataset loaded into SQL database successfully!")
```

```python
#Run SQL queries

# Step 10: Inspect Data Using SQL

query = "SELECT * FROM insurance LIMIT 10"

sql_result = pd.read_sql_query(query, conn)

display(sql_result)
```

```python
#Analyze the dataset using SQL

# Step 11: Count Records and Inspect Table Structure

query_count = """
SELECT COUNT(*) AS total_records
FROM insurance
"""

query_columns = "PRAGMA table_info(insurance)"

print("Total Records:")
display(pd.read_sql_query(query_count, conn))

print("SQL Table Columns:")
display(pd.read_sql_query(query_columns, conn))
```

```python
#SQL se dataset ka overview

# Step 12: Dataset Overview Using SQL

query = """
SELECT
    COUNT(*) AS total_records
FROM insurance
"""

display(pd.read_sql_query(query, conn))

query = "SELECT * FROM insurance LIMIT 5"
display(pd.read_sql_query(query, conn))
```

```python
#SQL se category-wise record count

# Step 13: Category-wise Analysis Using SQL

categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

for col in categorical_columns:
    print("\nCategory Analysis:", col)

    query = f'''
    SELECT "{col}", COUNT(*) AS total_records
    FROM insurance
    GROUP BY "{col}"
    ORDER BY total_records DESC
    LIMIT 15
    '''

    display(pd.read_sql_query(query, conn))
```

```python
#Numerical data analysis using SQL

# Step 14: Numerical Summary Using SQL

numeric_columns = df.select_dtypes(
    include="number"
).columns.tolist()

for col in numeric_columns:
    print("\nNumerical Analysis:", col)

    query = f'''
    SELECT
        COUNT("{col}") AS total_records,
        ROUND(AVG("{col}"), 2) AS average_value,
        ROUND(MIN("{col}"), 2) AS minimum_value,
        ROUND(MAX("{col}"), 2) AS maximum_value
    FROM insurance
    '''

    display(pd.read_sql_query(query, conn))
```

```python
#Find the most frequent values

# Step 15: Most Frequent Values

for col in categorical_columns:
    print("\nMost Frequent:", col)

    query = f'''
    SELECT "{col}", COUNT(*) AS frequency
    FROM insurance
    GROUP BY "{col}"
    ORDER BY frequency DESC
    LIMIT 5
    '''

    display(pd.read_sql_query(query, conn))
```

```python
#Check the cleaned dataset

# Step 16: Data Quality Check

print("Total Rows:", len(df))
print("Total Columns:", len(df.columns))
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
#SQL mein filtered records check karo

# Step 17: Filter Records Using SQL

if numeric_columns:
    col = numeric_columns[0]

    query = f'''
    SELECT *
    FROM insurance
    WHERE "{col}" IS NOT NULL
    LIMIT 20
    '''

    display(pd.read_sql_query(query, conn))
else:
    print("No numerical columns found.")
```

```python
#Sort records using SQL

# Step 18: Sort Records

if numeric_columns:
    col = numeric_columns[0]

    query = f'''
    SELECT *
    FROM insurance
    ORDER BY "{col}" DESC
    LIMIT 10
    '''

    display(pd.read_sql_query(query, conn))
else:
    print("No numerical columns found.")
```
